```python
dataset_url = 'https://raw.githubusercontent.com/SampattKumar/Linear-Regression-Car-Dataset/refs/heads/master/CarPrice_Assignment.csv'
```

```python
import pandas as pd
import numpy as np
```

```python
raw_df = pd.read_csv(dataset_url)
raw_df
```

```python
raw_df['brand'] = raw_df['CarName'].apply(lambda x: x.split(' ')[0])
raw_df['brand']
```

```python
X = raw_df.drop(['CarName', 'brand', 'price'], axis=1)
y = raw_df['price']
```

```python
raw_df.info()
```

```python
num_col = X.select_dtypes(include=np.number).columns.tolist()
cat_col = X.select_dtypes(include='str').columns.tolist()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.2, random_state=15)
```

```python
from sklearn.preprocessing import MinMaxScaler, OneHotEncoder
```

```python
scaler = MinMaxScaler()
```

```python
X_train[num_col] = scaler.fit_transform(X_train[num_col])
X_test[num_col] = scaler.transform(X_test[num_col])
```

```python
encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False,
)

X_train_encoded = encoder.fit_transform(X_train[cat_col])
X_test_encoded = encoder.transform(X_test[cat_col])
```

```python
encoded_col = encoder.get_feature_names_out(cat_col)

X_train_encoded_df = pd.DataFrame(
    X_train_encoded, columns=encoded_col, index=X_train.index
)
X_test_encoded_df = pd.DataFrame(
    X_test_encoded, columns=encoded_col, index=X_test.index
)
```

```python
X_train = pd.concat(
    [X_train.drop(columns=cat_col, errors='ignore'), X_train_encoded_df], axis=1
)
X_test = pd.concat(
    [X_test.drop(columns=cat_col, errors='ignore'), X_test_encoded_df], axis=1
)

X_train
```

```python
# X_test = X_test.reindex(columns=X_train.columns, fill_value=0)
```

```python
%%time
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
y_pred
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print(f"Training rows: {len(X_train)}")
print(f"Testing rows:  {len(X_test)}")
print(f"Mean Absolute Error (MAE): {mae:.2f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.2f}")
print(f"R² Score: {r2:.4f}")
```
